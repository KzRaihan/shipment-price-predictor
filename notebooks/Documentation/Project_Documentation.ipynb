#  Shipment Price Prediction — Linear Regression

## 1. Problem Statement

### Objective

The objective of this project is to build a Machine Learning system that predicts the **shipping cost required to deliver sculptures to customers** based on information about the sculpture, artist, shipment method, customer location, and delivery requirements.

The company sells sculptures created by artists from different parts of the world. Since shipping costs can vary depending on factors such as sculpture dimensions, weight, material, international delivery, express shipment, transportation method, fragile handling, remote location, and other shipment-related conditions, an accurate prediction system can help estimate shipping costs before an order is fulfilled.

### Input Features

The model utilizes information related to the sculpture, artist, shipment, and customer, including:

* Customer ID
* Artist Name
* Artist Reputation
* Height
* Width
* Weight
* Material
* Price of Sculpture
* Base Shipping Price
* International Shipment
* Express Shipment
* Installation Included
* Transport Method
* Fragile
* Customer Information
* Remote Location
* Scheduled Date
* Delivery Date
* Customer Location

### Target Variable

* **Shipping Cost (`Cost`)**

  * A continuous numerical value representing the estimated cost required to ship the sculpture.

# 2. Solution Overview

## Business Impact

A shipment price prediction system can help the company estimate delivery costs before completing an order.

The prediction system can potentially support:

* More accurate shipping cost estimation
* Improved pricing decisions
* Better shipment planning
* Faster quotation generation
* Identification of factors affecting shipping costs
* Reduction of manual cost estimation
* Improved operational planning

## Solution Approaches

### Approach 1 — Machine Learning Regression (Selected)

Since the target variable **`Cost`** is continuous, this project is formulated as a **Supervised Machine Learning Regression** problem.

Potential regression algorithms include:

* Linear Regression
* Ridge Regression
* Lasso Regression
* Elastic Net
* Decision Tree Regressor
* Random Forest Regressor
* Gradient Boosting Regressor
* AdaBoost Regressor
* XGBoost Regressor
* CatBoost Regressor
* LightGBM Regressor (if applicable)

Multiple regression algorithms can be trained and evaluated to identify a suitable model for predicting shipment costs.

### Approach 2 — Deep Learning

A neural network-based regression model can also be used to predict the continuous shipping cost.

However, because the dataset consists primarily of structured/tabular data, traditional Machine Learning regression algorithms can provide an effective baseline with lower computational complexity and easier interpretability.

**Selected Approach:** Machine Learning Regression

## Phase 1 — Experimentation (Jupyter Notebook)

The experimentation phase focuses on understanding the dataset, preparing the features, training different regression models, and selecting an appropriate model for production.

### Step 1: Data Collection

* Load the shipment dataset.
* Inspect the dataset structure.
* Identify numerical and categorical features.
* Analyze data types and feature information.
* Understand the target variable `Cost`.

### Step 2: Exploratory Data Analysis (EDA)

The EDA process focuses on understanding the factors that may influence shipment costs.

Activities include:

* Analyze dataset dimensions.
* Check missing values.
* Check duplicate records.
* Analyze numerical feature distributions.
* Analyze categorical feature distributions.
* Detect potential outliers.
* Analyze the distribution of shipping cost.
* Study relationships between features and shipping cost.
* Analyze correlations among numerical variables.
* Identify potentially important features.
* Visualize key patterns and relationships.

### Step 3: Data Preprocessing

Prepare the raw dataset for Machine Learning.

Activities include:

* Handle missing values.
* Remove or investigate duplicate records.
* Handle inconsistent data.
* Separate input features and target variable.
* Identify numerical and categorical columns.
* Encode categorical variables.
* Scale numerical features where required.
* Handle date-related features.
* Create a reusable preprocessing pipeline.

### Step 4: Feature Engineering

Create meaningful features that can improve the model's ability to predict shipment cost.

Potential feature engineering includes:

* Sculpture volume
* Sculpture dimensions
* Weight-related features
* Sculpture price-related features
* Shipment type features
* International shipment indicators
* Express shipment indicators
* Fragile shipment indicators
* Remote location indicators
* Delivery duration
* Scheduled-to-delivery time difference
* Artist reputation-based features
* Interaction between shipment-related variables

The final feature set should be determined through EDA and model experimentation rather than adding unnecessary variables.

### Step 5: Model Training

Train multiple regression algorithms using the prepared dataset.

Candidate models include:

* Linear Regression
* Ridge Regression
* Lasso Regression
* Elastic Net
* Decision Tree Regressor
* Random Forest Regressor
* Gradient Boosting Regressor
* AdaBoost Regressor
* XGBoost Regressor
* CatBoost Regressor

The models are trained using the same preprocessing and train-test split strategy to ensure a fair comparison.

### Step 6: Hyperparameter Tuning

Optimize the top-performing regression models using techniques such as:

* Grid Search
* Randomized Search
* Cross Validation

Hyperparameter tuning can be applied to models such as:

* Ridge
* Lasso
* Random Forest
* Gradient Boosting
* XGBoost
* CatBoost

### Step 7: Model Evaluation

Regression models are evaluated using appropriate regression metrics.

Key evaluation metrics include:

* **MAE — Mean Absolute Error**
* **MSE — Mean Squared Error**
* **RMSE — Root Mean Squared Error**
* **R² Score**
* **Adjusted R²** where applicable

The evaluation process should consider both prediction accuracy and the business meaning of the prediction error.

### Step 8: Model Selection

Select the final model based on experimental results and the project's requirements.

The selected model should provide:

* Strong predictive performance
* Reasonable generalization
* Appropriate computational cost
* Stable performance
* Suitable inference speed
* Maintainability for production deployment

# Phase 2 — Production ML Pipeline

After completing experimentation, the solution is converted into a modular and production-ready Machine Learning pipeline.

## Pipeline Components

### 1. Data Ingestion

* Read the raw shipment dataset.
* Connect to the required data source.
* Load the dataset.
* Split the data into training and testing datasets.
* Store datasets as artifacts.

### 2. Data Validation

Validate the quality and structure of incoming shipment data.

Activities include:

* Validate dataset schema.
* Check required columns.
* Validate data types.
* Check missing values.
* Detect unexpected values.
* Validate numerical ranges.
* Check data quality.
* Detect potential data drift.

### 3. Data Transformation

Transform raw data into a format suitable for model training.

Activities include:

* Handle missing values.
* Encode categorical features.
* Transform date features.
* Scale numerical features where required.
* Apply feature engineering.
* Create reusable preprocessing objects.
* Save the preprocessing pipeline as an artifact.

### 4. Model Trainer

* Load transformed training data.
* Train the selected regression model.
* Perform model training.
* Apply hyperparameters obtained during experimentation.
* Save the trained model as an artifact.

### 5. Model Evaluation

Evaluate the newly trained model against the test dataset.

Evaluation includes:

* Calculate MAE.
* Calculate MSE.
* Calculate RMSE.
* Calculate R².
* Compare with the baseline model.
* Validate whether the model meets the required performance threshold.
* Approve or reject the trained model.

### 6. Model Pusher

* Push the approved model to the deployment environment.
* Store the trained model.
* Version model artifacts.
* Store the preprocessing pipeline.
* Prepare the model for inference.
