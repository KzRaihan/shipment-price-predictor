### 3. Load the Dataset into MongoDB Atlas (cloud)

```python
# import library
import pymongo
```

```python
# Database name
DB_NAME = "Shipment_Kz"
COLLECTION_NAME = "shipment_collection"
CONNECTION_URL = "Your_MONGO_BD_CONNECTION_URL"
```

- short note:

    - DB_NAME = Database name

    - COLLECTION_NAME = Table name of the Database

    - CONNECTION_URL = Connection string between MongoDB Atlas and python

### Storing Data to MongoDB Database

```python
client = pymongo.MongoClient(CONNECTION_URL)
data_base = client[DB_NAME]
collection = data_base[COLLECTION_NAME]
```

```python
rec = collection.insert_many(data_dic)
```

### Fetching Data from MongoDB Database

```python
# Fetching all records
records = collection.find()
```

```python
# it give us an object of the data and it can be iterable
records
```

```python
# one one to display
# for i in records:
#     print(i)
```

### Convert records into DataFrame

```python
# fetching all record from the MongoDB atlas and convert into dataframe
df = pd.DataFrame(list(collection.find()))
```

```python
# show first 5 rows
df.head()
```

```python
# Drop the _id columns
if "_id" in df.columns:
    df.drop(columns="_id", inplace=True)
```

- Key Note:
    - Because, MongoDB Create a unique Id for a dataset.

```python
df.head()
```

#### Convert DataFrame to CSV format

```python
# convert DataFrame -> CSV
df.to_csv("shipment.csv", index = False)
```
